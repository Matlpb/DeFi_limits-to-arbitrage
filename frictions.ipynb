```python
import pandas as pd

# Read both CSVs
df_pancake_swap = pd.read_csv('ethereum/swaps/df_pancake_swap.csv')
df_pancake_gas = pd.read_csv('ethereum/gas/pancake_gas_price.csv')

```

```python
df_pancake_swap.columns = df_pancake_swap.columns.str.strip()
df_pancake_gas.columns = df_pancake_gas.columns.str.strip()

# Ensure block numbers are numeric
df_pancake_swap['evt_block_number'] = pd.to_numeric(
    df_pancake_swap['evt_block_number'], errors='coerce'
)
df_pancake_gas['evt_block_number'] = pd.to_numeric(
    df_pancake_gas['evt_block_number'], errors='coerce'
)

# Sort by block number
df_pancake_swap = df_pancake_swap.sort_values('evt_block_number')
df_pancake_gas = df_pancake_gas.sort_values('evt_block_number')

print("SWAP")
print("rows:", len(df_pancake_swap))
print("min block:", df_pancake_swap['evt_block_number'].min())
print("max block:", df_pancake_swap['evt_block_number'].max())

print("\nGAS")
print("rows:", len(df_pancake_gas))
print("min block:", df_pancake_gas['evt_block_number'].min())
print("max block:", df_pancake_gas['evt_block_number'].max())



swap_blocks = set(df_pancake_swap['evt_block_number'])
gas_blocks = set(df_pancake_gas['evt_block_number'])

print("Only in swap:")
print(sorted(swap_blocks - gas_blocks))

print("\nOnly in gas:")
print(sorted(gas_blocks - swap_blocks))
```

```python
missing_in_gas = swap_blocks - gas_blocks
missing_in_swap = gas_blocks - swap_blocks

print("Rows in swap whose block number is not in gas:")
display(
    df_pancake_swap[
        df_pancake_swap['evt_block_number'].isin(missing_in_gas)
    ]
)

print("Rows in gas whose block number is not in swap:")
display(
    df_pancake_gas[
        df_pancake_gas['evt_block_number'].isin(missing_in_swap)
    ]
)
```

```python
keys = ['evt_block_number', 'evt_tx_hash', 'pool']



duplicates = df_pancake_swap[
    df_pancake_swap.duplicated(subset=keys, keep=False)
].sort_values(keys)

print(f"Duplicate rows: {len(duplicates)}")
display(duplicates))
```

```python
import pandas as pd

# Read both CSVs
df_pancake_swap = pd.read_csv('ethereum/swaps/df_pancake_swap.csv')
df_pancake_gas = pd.read_csv('ethereum/gas/pancake_gas_price.csv')

# Strip whitespace from column names
df_pancake_swap.columns = df_pancake_swap.columns.str.strip()
df_pancake_gas.columns = df_pancake_gas.columns.str.strip()

# Find duplicate rows in swap on the join keys
join_keys = ['evt_block_number', 'evt_tx_hash', 'pool']
duplicates_swap = df_pancake_swap[df_pancake_swap.duplicated(subset=join_keys, keep=False)]

print(f"Total duplicate rows in swap: {len(duplicates_swap)}")
print("\nDuplicate rows in swap (sorted by keys):")
print(duplicates_swap[join_keys + ['evt_block_time', 'fee', 'liquidity', 'mid_price']].sort_values(join_keys))

# Show which rows are in swap but not in gas (by key)
swap_keys = set(df_pancake_swap[join_keys].apply(tuple, axis=1))
gas_keys = set(df_pancake_gas[join_keys].apply(tuple, axis=1))

rows_in_swap_not_in_gas = swap_keys - gas_keys
print(f"\n\nRows in SWAP but NOT in GAS: {len(rows_in_swap_not_in_gas)}")
if rows_in_swap_not_in_gas:
    mask = df_pancake_swap[join_keys].apply(tuple, axis=1).isin(rows_in_swap_not_in_gas)
    print(df_pancake_swap[mask][join_keys + ['evt_block_time', 'fee', 'liquidity']])
```
